```python
import pandas as pd

# Load datasets
diabetes = pd.read_csv("datasets/diabetes.csv")
heart = pd.read_csv("datasets/heart.csv")
parkinsons = pd.read_csv("datasets/parkinsons.csv")

# Check data
print("Diabetes shape:", diabetes.shape)
print("Heart shape:", heart.shape)
print("Parkinsons shape:", parkinsons.shape)
```

```python
# Display first 5 rows of each dataset
print("Diabetes dataset:")
display(diabetes.head())

print("\nHeart dataset:")
display(heart.head())

print("\nParkinsons dataset:")
display(parkinsons.head())
```

```python
# Check column names, data types, missing values
print("Diabetes info:")
diabetes.info()

print("\nHeart info:")
heart.info()

print("\nParkinsons info:")
parkinsons.info()

```

```python
# Fill numeric columns with median
heart.fillna(heart.median(numeric_only=True), inplace=True)

# Fill object columns with mode
for col in heart.select_dtypes(include='object').columns:
    heart[col].fillna(heart[col].mode()[0], inplace=True)
```

```python
print(parkinsons.columns)

```

```python
# Drop the first column by position
parkinsons.drop(parkinsons.columns[0], axis=1, inplace=True)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, confusion_matrix

# Features & target
X_diabetes = diabetes.drop('Outcome', axis=1)
y_diabetes = diabetes['Outcome']

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(X_diabetes, y_diabetes, test_size=0.2, random_state=42)

# Model
model = RandomForestClassifier(random_state=42)
model.fit(X_train, y_train)

# Prediction
y_pred = model.predict(X_test)

# Evaluation
print("Diabetes Accuracy:", accuracy_score(y_test, y_pred))
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))
```

```python
# Convert target to binary: 0 = No disease, 1 = Disease
heart['num'] = heart['num'].apply(lambda x: 0 if x==0 else 1)

# Encode object columns
heart_encoded = pd.get_dummies(heart.drop('id', axis=1), drop_first=True)

# Features & target
X_heart = heart_encoded.drop('num', axis=1)
y_heart = heart_encoded['num']

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(X_heart, y_heart, test_size=0.2, random_state=42)

# Model
model = RandomForestClassifier(random_state=42)
model.fit(X_train, y_train)

# Prediction
y_pred = model.predict(X_test)

# Evaluation
print("Heart Disease Accuracy:", accuracy_score(y_test, y_pred))
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))
```

```python
# Features & target
X_parkinsons = parkinsons.drop('status', axis=1)
y_parkinsons = parkinsons['status']

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(X_parkinsons, y_parkinsons, test_size=0.2, random_state=42)

# Model
model = RandomForestClassifier(random_state=42)
model.fit(X_train, y_train)

# Prediction
y_pred = model.predict(X_test)

# Evaluation
print("Parkinsons Accuracy:", accuracy_score(y_test, y_pred))
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))
```
